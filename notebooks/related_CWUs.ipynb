```python
from rdflib import Graph

g = Graph()
g.parse("data8.ttl", format="turtle")

q = """
PREFIX ns1: <https://w3id.org/write/ontology#>
CONSTRUCT {
  ?u1 ns1:relatedUnit ?u2 .
  ?u2 ns1:relatedUnit ?u1 .
}
WHERE {
  ?artwork ns1:hasUnit ?u1 .
  ?artwork ns1:hasUnit ?u2 .
  FILTER(?u1 != ?u2)
}
"""

inferred = g.query(q)
for s, p, o in inferred:
    if (s, p, o) not in g:
        g.add((s, p, o))

g.serialize("data9.ttl", format="turtle", encoding="utf-8", indent=4)
```
