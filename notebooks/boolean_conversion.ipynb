```python
import json

# Sample JSON-LD input
input_file = "data3.jsonld"
output_file = "data4.jsonld"

# Mapping rules
mapping = {
    "Yes": True,
    "No": False,
    "Present": True,
    "Not present": False
}

# Properties to transform
boolean_properties = [
    "write:hasCalligraphicLine",
    "write:hasColophon",
    "write:hasSeal",
    "write:hasChengyu",
    "write:hasBigCharactersCalligraphy",
    "write:vectorisedOrDigitised"
]

def transform_boolean_properties(obj):
    if isinstance(obj, dict):
        for key, value in obj.items():
            if key in boolean_properties and isinstance(value, list):
                for item in value:
                    if isinstance(item, dict) and "@value" in item:
                        original = item["@value"]
                        item["@value"] = mapping.get(original, None)
                        item["@type"] = "xsd:boolean"
                        item.pop("@language", None)  # Remove @language if present
                        item.pop("type", None)
            else:
                transform_boolean_properties(value)
    elif isinstance(obj, list):
        for item in obj:
            transform_boolean_properties(item)

# Load JSON-LD file
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

# Transform properties
transform_boolean_properties(data)

# Save transformed JSON-LD
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print(f"Transformed JSON-LD saved to {output_file}")
```
