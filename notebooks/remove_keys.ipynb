```python
import json

# List of keys to remove
REMOVE_KEYS = [
    "value_resource_id",
    "value_resource_name",
    "url",
    "thumbnail_url",
    "thumbnail_title",
    "thumbnail_type",
    "type",
    "property_label"
]

def remove_unwanted_keys(obj):
    """Recursively remove keys in REMOVE_KEYS from dicts/lists."""
    if isinstance(obj, dict):
        return {k: remove_unwanted_keys(v) for k, v in obj.items() if k not in REMOVE_KEYS}
    elif isinstance(obj, list):
        return [remove_unwanted_keys(i) for i in obj]
    else:
        return obj

# Load JSON-LD file
with open("data2.jsonld", "r", encoding="utf-8") as f:
    data = json.load(f)

# Clean the data
cleaned_data = remove_unwanted_keys(data)

# Save cleaned JSON-LD
with open("data3.jsonld", "w", encoding="utf-8") as f:
    json.dump(cleaned_data, f, ensure_ascii=False, indent=2)
```
